```python
file_path = '/content/heart (3).csv'
df = pd.read_csv(file_path)
print(df)
```

```python
x = df.drop(columns=['target'])
y = df['target']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
import numpy as np
```

```python
def custom_feature_selection(X):
    num_features = X.shape[1]
    selected_features = np.random.choice(num_features, size=num_features//2, replace=False)
    print("Selected Features Indices:", selected_features)  #prints selected feature indices
    assert len(selected_features) == num_features // 2  #ensures correct number of features are selected
    selected_features_names = X.columns[selected_features]
    return X[selected_features_names]
```

```python
x_train_selected = custom_feature_selection(pd.DataFrame(x_train_scaled))
x_test_selected = custom_feature_selection(pd.DataFrame(x_test_scaled))
```

```python
import tensorflow as tf
```

```python
# Defining custom weight ftn
def custom_weight_init(shape, dtype=None):
    # Initializing weight using random normal distribution
    return tf.Variable(tf.random.normal(shape, stddev=0.01), dtype=dtype)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
# Designing out CNN model
model = Sequential([
    Dense(64, input_dim=x_train_selected.shape[1], activation='sigmoid', kernel_initializer=custom_weight_init),
    Dense(64, activation='sigmoid', kernel_initializer=custom_weight_init),
    Dense(64, activation='sigmoid', kernel_initializer=custom_weight_init),
    Dense(64, activation='sigmoid', kernel_initializer=custom_weight_init),
    Dense(64, activation='sigmoid', kernel_initializer=custom_weight_init),
    Dense(1, activation='sigmoid', kernel_initializer=custom_weight_init)
])
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
model.fit(x_train_selected, y_train, epochs=10, batch_size=32, validation_split=0.2)
```
